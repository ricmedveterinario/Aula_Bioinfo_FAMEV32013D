```python
#1.Comparação entre Frequências de DNA em Diferentes Espécies
# Definição: A frequência de nucleotídeos indica quantas vezes cada base (A, T, C, G) aparece em uma sequência de DNA. Isso é útil para avaliar o conteúdo GC, AT, detectar anomalias e entender a composição genética.
# Comparação de Frequência de Nucleotídeos: Bactéria vs Planta vs Animal

from collections import Counter
import matplotlib.pyplot as plt

# Sequência simulada de bactéria (ex: *E. coli* – mais AT)
seq_bacteria = "ATGCGTATATCGATATCGATATAGCGTATATATCG"

# Sequência simulada de planta (ex: *Arabidopsis thaliana* – equilibrada)
seq_planta = "ATGCCGCGCGTACGCGGCGCGTATGCGCGCGTACG"

# Sequência simulada de animal (ex: *Mus musculus* – mamífero com GC moderado)
seq_animal = "ATGCGCGTACGCGTATAGCGTACGCGTATAGCGTA"

def calcular_frequencia(seq):
    contagem = Counter(seq)
    total = len(seq)
    return [contagem[base] / total for base in "ATCG"]

freq_bact = calcular_frequencia(seq_bacteria)
freq_planta = calcular_frequencia(seq_planta)
freq_animal = calcular_frequencia(seq_animal)

# Plot
import numpy as np

bases = ['A', 'T', 'C', 'G']
x = np.arange(len(bases))
largura = 0.25

plt.figure(figsize=(9, 5))
plt.bar(x - largura, freq_bact, width=largura, label=' Bactéria (E. coli)', color='skyblue')
plt.bar(x, freq_planta, width=largura, label=' Planta (A. thaliana)', color='seagreen')
plt.bar(x + largura, freq_animal, width=largura, label=' Animal (Mus musculus)', color='salmon')

plt.xticks(x, bases)
plt.ylabel("Proporção (%)")
plt.title("Comparação da Frequência de Nucleotídeos")
plt.ylim(0, 0.6)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


```
