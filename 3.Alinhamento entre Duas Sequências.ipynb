```python
#3.Gráfico de Alinhamento entre Duas Sequências

import matplotlib.pyplot as plt

# Mesmas sequências
ref =   "ATGCGTACGTTAGCGATCGATCGTACGATCGTAA"
query = "ATGCGTACGATAGCAATGGATCGTACGCTCGTAA"

cores = []
for r, q in zip(ref, query):
    if r == q:
        cores.append("green")  # igual
    else:
        cores.append("orange")  # diferente

plt.figure(figsize=(10, 1))
plt.bar(range(len(ref)), [1]*len(ref), color=cores, edgecolor="black")
plt.title("Alinhamento: Verde = Igual, Laranja = Mutação")
plt.xticks(range(0, len(ref), 2))
plt.yticks([])
plt.tight_layout()
plt.show()
```
