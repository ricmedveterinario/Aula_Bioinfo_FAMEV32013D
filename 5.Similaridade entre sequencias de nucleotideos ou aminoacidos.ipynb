```python
#5. Similaridade
# Definição: Considera bases quimicamente semelhantes (importante para proteínas); no DNA, similaridade = identidade, mas em proteínas ela é mais flexível.
# No exemplo abaixo, L ↔ F são ambos hidrofóbicos, portanto considerados "similares".

import matplotlib.pyplot as plt

seq1 = "MKTLLIA"
seq2 = "MKTLFIA"

# Pares considerados similares (exemplo simples)
pares_similares = [('L','F'), ('F','L')]

posicoes = []
cores = []

for i, (a, b) in enumerate(zip(seq1, seq2)):
    if a == b:
        cores.append('green')
    elif (a, b) in pares_similares:
        cores.append('orange')
    else:
        cores.append('red')
    posicoes.append(i)

# Plot
plt.figure(figsize=(8,2))
plt.bar(posicoes, [1]*len(posicoes), color=cores)
plt.xticks(posicoes, [f"{a}|{b}" for a,b in zip(seq1,seq2)])
plt.title("Similaridade entre proteínas")
plt.yticks([])
plt.tight_layout()
plt.show()

# Cálculo
similares = sum(1 for a, b in zip(seq1, seq2) if a == b or (a,b) in pares_similares)
print(f"Similaridade: {similares/len(seq1):.2%}")

```
