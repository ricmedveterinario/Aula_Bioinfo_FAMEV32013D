```python
# Uma matriz de substituição
# Definição: Usa uma matriz de substituição simples para criarmos uma pontuação de alinhamento
# match = +1 / +1 ponto para cada posição com aminoácidos iguais (match ✅)
# mismatch = -1 / –1 ponto para cada posição com aminoácidos diferentes (mismatch ❌)
# É uma medida numérica da semelhança entre duas sequências de mesmo comprimento, baseada em:

import matplotlib.pyplot as plt

seq1 = "MKTLLIA"
seq2 = "MKTLFIA"

match = 1
mismatch = -1

score_total = 0
valores = []

for a, b in zip(seq1, seq2):
    if a == b:
        valores.append(match)
        score_total += match
    else:
        valores.append(mismatch)
        score_total += mismatch

plt.figure(figsize=(8,2))
plt.bar(range(len(seq1)), valores, color=['green' if v==1 else 'red' for v in valores])
plt.xticks(range(len(seq1)), [f"{a}|{b}" for a,b in zip(seq1,seq2)])
plt.title(f"Score de Alinhamento: {score_total}")
plt.axhline(0, color='black', linestyle='--')
plt.tight_layout()
plt.show()
```
