```python
#4.Porcentagem de Identidade entre Alinhamentos
#Definição: Proporção de posições alinhadas idênticas entre duas sequências.
import matplotlib.pyplot as plt

# Simular identidade entre várias amostras
sequencias = ["Amostra 1", "Amostra 2", "Amostra 3", "Amostra 4"]
identidades = [98.5, 87.0, 92.3, 76.4]

plt.figure(figsize=(8, 4))
plt.bar(sequencias, identidades, color="purple")
plt.ylabel("Identidade (%)")
plt.ylim(0, 100)
plt.title("Porcentagem de identidade com a sequência de referência")
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
