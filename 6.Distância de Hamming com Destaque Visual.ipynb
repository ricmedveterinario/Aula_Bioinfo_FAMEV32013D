```python
#Distância de Hamming com Destaque Visual
#Definição: Conta quantas posições são diferentes entre duas sequências do mesmo tamanho.

import matplotlib.pyplot as plt

seq1 = "GATTACA"
seq2 = "GACTATA"

hamming = sum(1 for a, b in zip(seq1, seq2) if a != b)

# Gráfico
cores = ['red' if a != b else 'green' for a, b in zip(seq1, seq2)]

plt.figure(figsize=(8, 2))
plt.bar(range(len(seq1)), [1]*len(seq1), color=cores)
plt.xticks(range(len(seq1)), [f"{a}|{b}" for a,b in zip(seq1,seq2)])
plt.title(f"Distância de Hamming = {hamming}")
plt.yticks([])
plt.tight_layout()
plt.show()
```
