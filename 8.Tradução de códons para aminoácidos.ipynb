```python
#8.Tradução de códons para aminoácidos
#Definição: O DNA é lido em trincas (códons), traduzidas em aminoácidos.

# Dicionário simplificado
import matplotlib.pyplot as plt

# Dicionário de tradução simplificado
tabela_codons = {
    "ATG": "Met", "TGG": "Trp", "TAA": "Stop", "TGA": "Stop", "TAG": "Stop",
    "GTT": "Val", "GTC": "Val", "GTA": "Val", "GTG": "Val"
}

# Sequência de DNA de exemplo
seq_dna = "ATGGTTGTCGTAATGTAA"

# Variáveis para armazenar códons e aminoácidos
codons = []
aminoacidos = []

# Tradução
for i in range(0, len(seq_dna), 3):
    codon = seq_dna[i:i+3]
    if len(codon) < 3:
        print(f"{codon} → (incompleto)")
        break
    aa = tabela_codons.get(codon, "???")
    codons.append(codon)
    aminoacidos.append(aa)
    if aa == "Stop":
        break

# Mostrar a tradução
print("Tradução:")
for c, a in zip(codons, aminoacidos):
    print(f"{c} → {a}")

# 🎨 Visualização
cores = ['green' if a != 'Stop' else 'red' for a in aminoacidos]

plt.figure(figsize=(10, 2))
plt.bar(range(len(codons)), [1]*len(codons), color=cores, tick_label=codons)
for i, a in enumerate(aminoacidos):
    plt.text(i, 1.05, a, ha='center', fontsize=10)
plt.ylim(0, 1.5)
plt.title("Tradução de DNA → Aminoácidos (códons)")
plt.axhline(0, color='black')
plt.tight_layout()
plt.show()

```
