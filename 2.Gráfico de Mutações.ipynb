```python
#2.Gráfico de Mutações (por posição)
import matplotlib.pyplot as plt

# Simulação de duas sequências alinhadas
ref =   "ATGGCTACGTAGCGCATGCGTAGCATCGATCGTAA"
mutada = "ATGGCTACGATAGCAATGGATCGTACGCTCGTAAA"

# Detectar diferenças
posicoes = []
bases_ref = []
bases_mut = []

for i, (r, m) in enumerate(zip(ref, mutada)):
    if r != m:
        posicoes.append(i)
        bases_ref.append(r)
        bases_mut.append(m)

# Plotagem
plt.figure(figsize=(10, 3))
plt.scatter(posicoes, [1]*len(posicoes), color='red', s=100)

# Adiciona os textos das mutações
for x, ref_base, mut_base in zip(posicoes, bases_ref, bases_mut):
    plt.text(x, 1.05, f"{ref_base}→{mut_base}", ha='center', fontsize=9)

plt.yticks([])
plt.xlabel("Posição na sequência")
plt.title("Mutações em sequência de DNA")
plt.grid(True)
plt.tight_layout()
plt.show()

# Imprimir detalhes das mutações
for i, r, m in zip(posicoes, bases_ref, bases_mut):
    print(f"Posição {i+1}: {r} → {m}")

```
